**1) BAR CHART** : **Which product category brings in the most sales ?**

```python
category_sales = data.groupby("Category")["Sales"].sum().sort_values(ascending=False).reset_index()
print(category_sales)

plt.figure(figsize=(8, 5))
sns.barplot(data=category_sales, x="Category", y="Sales", hue="Category", palette="coolwarm", legend=False)
plt.title("Total Sales by Product Category", fontsize=14, fontweight="bold")
plt.xlabel("Category",fontsize=12, fontweight="bold")
plt.ylabel("Total Sales",fontsize=12, fontweight="bold")
plt.tight_layout()
plt.savefig("1_bar_category_sales.png", dpi=300)
plt.show()
```

**BOX PLOT** : **How does order value differ across product sub-categories?**

```python
plt.figure(figsize=(12, 7))
sns.boxenplot(x="Region", y="Sales", hue="Category", data=data)
plt.yscale("log")
plt.title("Order Value Distribution by Region and Category", fontsize=14, fontweight="bold")
plt.xlabel("Region",fontsize=12, fontweight="bold")
plt.ylabel("Sales per Order (log scale)",fontsize=12, fontweight="bold")
plt.legend(title="Category", loc="upper left", bbox_to_anchor=(1, 1))
plt.savefig("2_box_sales_by_region_category.png", dpi=300, bbox_inches="tight")
plt.show()
```

heatmap : Which product sub-categories make or lose the most profit in each region?

```python
heat = data.pivot_table(values="Profit", index="Sub-Category", columns="Region", aggfunc="sum")

plt.figure(figsize=(9, 9))
sns.heatmap(heat, annot=True, fmt=".0f", center=0, linewidths=0.5)
plt.title("Total Profit by Sub-Category and Region", fontsize=14, fontweight="bold")
plt.xlabel("Region",fontsize=12, fontweight="bold")
plt.ylabel("Sub-Category",fontsize=12, fontweight="bold")
plt.savefig("3_heatmap_subcategory_region.png", dpi=300, bbox_inches="tight")
plt.show()
```

4. Line chart (How do sales change over time?)

```python
data["Order Date"] = pd.to_datetime(data["Order Date"], dayfirst=True)
data["Month"] = data["Order Date"].dt.to_period("M").astype(str)

monthly = data.groupby("Month")["Sales"].sum().reset_index()

plt.figure(figsize=(12, 5))
plt.plot(monthly["Month"], monthly["Sales"], color="red", marker="o", linewidth=2)
plt.xticks(rotation=90)
plt.title("Monthly Sales Over Time", fontsize=14, fontweight="bold")
plt.xlabel("Month",fontsize=12, fontweight="bold")
plt.ylabel("Total Sales",fontsize=12, fontweight="bold")
plt.savefig("4_line_monthly_sales.png", dpi=300, bbox_inches="tight")
plt.show()
```

Scatter plot (Do bigger discounts reduce profit?)

```python
plt.figure(figsize=(8, 5))
sns.scatterplot(x="Discount", y="Profit", data=data, hue="Category", palette="Set1")
plt.title("Discount vs Profit by Category", fontsize=14, fontweight="bold")
plt.xlabel("Discount",fontsize=12, fontweight="bold")
plt.ylabel("Profit",fontsize=12, fontweight="bold")
plt.savefig("5_scatter_discount_profit.png", dpi=300, bbox_inches="tight")
plt.show()
```
